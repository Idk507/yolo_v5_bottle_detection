````
import torch
from PIL import Image

def detect_objects(image_path, model_path='runs/train/exp/weights/last.pt'):
    model = torch.hub.load('ultralytics/yolov5', 'custom', path=model_path)
    img = Image.open(image_path)
    results = model(img)
    num_objects = len(results.pandas().xyxy[0])
    return num_objects

if __name__ == '__main__':
    image_path = '/content/frame00471.png'
    num_objects = detect_objects(image_path)
    print(f"Number of objects detected: {num_objects}")
````

````
import torch

model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)

import cv2

def count_objects_in_video(video_path):
    cap = cv2.VideoCapture(video_path)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    object_count = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        results = model(frame)


        object_count += len(results.pred[0])

    cap.release()
    return object_count


video_path = '/content/cam5.avi'
total_objects = count_objects_in_video(video_path)
print(f"Total objects detected in the video: {total_objects}")
````

````
import torch
import cv2
import numpy as np

# Load YOLOv5 model
model = torch.hub.load('ultralytics/yolov5', 'yolov5s')

# Open video capture
video_path = '/content/cam5test.avi'
cap = cv2.VideoCapture(video_path)

# Get video properties
frame_width = int(cap.get(3))
frame_height = int(cap.get(4))
fps = int(cap.get(5))

# Define the codec and create VideoWriter object
out = cv2.VideoWriter('output_video.mp4', cv2.VideoWriter_fourcc(*'XVID'), fps, (frame_width, frame_height))

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Perform object detection
    results = model(frame)

    # Get the detected classes and their counts
    detected_classes = results.pred[0][:, -1].cpu().numpy().astype(int)
    class_counts = {class_id: np.sum(detected_classes == class_id) for class_id in np.unique(detected_classes)}

    # Draw class counts on the frame
    font = cv2.FONT_HERSHEY_SIMPLEX
    font_scale = 1.0
    font_color = (0, 255, 0)
    font_thickness = 2
    y_coordinate = 30

    for class_id, count in class_counts.items():
        class_name = str(class_id)  # Replace with actual class names if available
        text = f'{class_name}: {count} objects'
        cv2.putText(frame, text, (10, y_coordinate), font, font_scale, font_color, font_thickness, cv2.LINE_AA)
        y_coordinate += 30

    # Write the frame to the output video
    out.write(frame)

    # Display the frame with counts
    cv2.imshow('Frame', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release video capture and writer
cap.release()
out.release()

# Close all OpenCV windows
cv2.destroyAllWindows()
````
